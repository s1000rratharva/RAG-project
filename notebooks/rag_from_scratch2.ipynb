```python
import os
from pypdf import PdfReader
from pathlib import Path
```

```python
pdf_path = Path(
    "C:/Ijjat me learning/RAG Project/documents/machine_learning.pdf"
)
reader=PdfReader(pdf_path)

print("Total Pages : ", len(reader.pages))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter=RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100,
    separators=["\n\n", "\n", " ", ""]
)
```

```python
documents=[]
metadata=[]
ids=[]

for page_no, page in enumerate(reader.pages):

    page_text=page.extract_text()

    if not page_text:
        continue

    page_chunks=text_splitter.split_text(page_text)

    for chunk_number, chunk in enumerate(page_chunks):

        documents.append(chunk)

        metadata.append({
            "Source": Path(pdf_path).name,
            "Page": page_no
        })

        ids.append(
            f"Page_{page_no}_chunk_{chunk_number}"
        )
    
```

```python
print(len(documents))
```

```python
from sentence_transformers import SentenceTransformer

model=SentenceTransformer("all-MiniLM-L6-v2")
```

```python
chunk_embedding=model.encode(
    documents,
    show_progress_bar=True
)

print("Total Chunks: ", len(documents))
print("Embedding shape:", chunk_embedding.shape)
```
